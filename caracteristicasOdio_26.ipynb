```python
#Uso Regex para limpiar tildes, diéresis, ñ y otros caractéres especiales que fueron cargados pero mal codificados 
#Declarar uso de IA
import re
import unicodedata 
def limpiar_texto(texto):
    # Diccionario manual de reemplazo de caracteres erróneos
    reemplazos = {
        'í±': 'ñ',
        'Ã\x83Â±': 'ñ',
        'í\xad': 'í',
        'í\xa1': 'á',
        'Ã¢Â\x80Â\x9c': '"',
        'Ã¢Â\x80Â\x9d': '"',
        'í©': 'é',
        'í³': 'ó',
        'íº': 'ú',
        'í§': 'ç',
        'í¼': 'ü' ,
        'ÃÂ¿': '¿',
        'Ã\x82Â¿': '¿',
        'Ã\x82Âº': 'º',  
        'Ã\x83Â©':'é',
        'ÃÂº': 'º',          
        'Ã\x82Âº': 'º',
        'Ã¡': 'á', 'Ã©': 'é', 'Ã­': 'í', 'Ã³': 'ó', 'Ãº': 'ú',
        'Ã\x82Â\xa0': ' ', 'Â\xa0': ' ', '\xa0': ' ',
        'ÃÂ¿': '¿', 'Ã\x82Â¿': '¿', 'Â¿': '¿',
        'ÃÂ¡': '¡', 'Ã\x82Â¡': '¡', 'Â¡': '¡',
        'Ã\x82Âº': 'º', 'ÃÂº': 'º', 'Âº': 'º',  # Indicador ordinal masculino (nº)
        'Ã\x82Âª': 'ª', 'ÃÂª': 'ª', 'Âª': 'ª',  # Indicador ordinal femenino (1ª)
        'Ã¢Â\x80Â\x9c': '"', 'Ã¢Â\x80Â\x9d': '"', # Comillas tipográficas dobles (“ ”)
        'Ã¢Â\x80Â\x98': "'", 'Ã¢Â\x80Â\x99': "'", # Comillas tipográficas simples (‘ ’)
        'Ã¢Â\x80Â¦': '...',                      # Puntos suspensivos especiales (…)
        'Ã¢Â\x82Â¬': '€' 
    }
    for error, correccion in reemplazos.items():
        texto = texto.replace(error, correccion)
    
    # Limpieza de caracteres de control, espacios y puntos consecutivos
    texto = unicodedata.normalize('NFC', texto)
    texto = re.sub(r'[\x00-\x1f\x7f-\x9f\xad]', '', texto)
    texto = re.sub(r'\s+', ' ', texto).strip()
    texto = re.sub(r'\.{2,}', ' ', texto)  
    
    return texto
 #A la misma columna, a cada registro se le aplica la función de limpieza limpiar_texto    
data['CONTENIDO A ANALIZAR'] = [limpiar_texto(t) for t in data['CONTENIDO A ANALIZAR']]
#Se imprime una muestra de la columna para verificar el texto limpio 
print(data['CONTENIDO A ANALIZAR'][20:50].to_list())


```

```python

#Contador de palabras
def contar_palabras_correctas(df):
   #Inicio un contador de palabras en 0
    total_palabras=0 
    #guardo la columna CONTENIDO A ANALIZAR en una variable llamada texto.
    texto=df['CONTENIDO A ANALIZAR'].astype(str).tolist()
    #Un bucle para cada documento preprocesado 
    for doc in nlp.pipe(texto, disable=["ner","parser"]):
        #Creo una lista para determinar si un token es válida(sin tener en cuenta signos de puntuación, sin espacios, sin stop words)
        palabras_valida=[token for token in doc 
                         if not token.is_punct 
                         and not token.is_space 
                         and not token.is_stop 
                         ]
        #Cada vez que tengo una palabra válida, incremento el contador.
        total_palabras=total_palabras + len(palabras_valida)
    return total_palabras
```

```python
# Aplico la función recién creada al dataset data para determinar el total de palabras.
total_general = contar_palabras_correctas(data)

print(f"Total de palabras en todo el dataset: {total_general}")
```

```python
# Creación de función para calcular el promedio de  oraciones en el dataframe a aplicar.
def promedio_oraciones(df):
    total_comentarios=len(df)
    total_oraciones=0
    datos = df['CONTENIDO A ANALIZAR'].astype(str).tolist()
    for doc in nlp.pipe(datos,disable=['ner']):
         oracion_comentario=len(list(doc.sents))
         total_oraciones += oracion_comentario
    promedio=total_oraciones/total_comentarios
    return promedio       
```

```python
# crear funcion de % ner
def pct_ner(df):
    datos=df['CONTENIDO A ANALIZAR'].astype(str).tolist()
    conteo_ner=0
    for doc in nlp.pipe(datos,disable=["parser"]):
        if len(doc.ents)>0:
            conteo_ner+=1
    return (conteo_ner/len(df))*100      
    
```

```python
# Calculamos los porcentajes
porcentaje_odio = pct_ner(df_odio)
porcentaje_no_odio = pct_ner(df_no_odio)

print(f"Resultados de Análisis NER:")
print(f"- Mensajes con ODIO que contienen entidades: {porcentaje_odio:.2f}%")
print(f"- Mensajes SIN ODIO que contienen entidades: {porcentaje_no_odio:.2f}%")
```

Donde un alto TTR sugiere un uso diverso del vocabulario y un TTR bajo sugiere uso repetido de terminos y un vocabulario limitado. Ejemplo: "El gato come, el gato duerme, el gato es negro.
```python
# Defino función que será aplicada a df_odio y df_no_odio
def adjetivos_odio(df):
    textos=df['CONTENIDO A ANALIZAR']
    #Esta lista guardará la totalidad de adjetivos encontrados
    lista_adjetivos=[]
    #Creo un bucle para cada documento del texto preprocesado
    for doc in nlp.pipe(textos):
        #Para cada token de ese documento, si el pos es adjetivo y no es puntuación ni espacio, lo lemmatiza y lo agrega a la lista
        for token in doc:
            if token.pos_=='ADJ' and not token.is_punct and not token.is_space and not token.is_stop:
                adjetivo=token.lemma_.lower()
                if len(adjetivo)>2:
                    lista_adjetivos.append(adjetivo)
    #La lista la convierto a serie para poder utilizar value_counts y tomar los top 15 basado en ese conteo.
    serie_adjetivos=pd.Series(lista_adjetivos)
    top_15 = serie_adjetivos.value_counts().head(15)
    return top_15
```

<span style="font-size: 14pt; font-weight: bold; color: #0098cd;">Dentro del corpus de odio, cuál es el porcentaje de paalabras escirtas en Mayúscula, con signos de exclamacion?</span>

```python
def agresividad(df):
    textos=df['CONTENIDO A ANALIZAR'].astype(str).tolist()
    total_palabras=0
    total_mayusculas=0
    total_tokens=0
    total_exclamaciones=0   
    signos=['!','¡','?','¿']
    

    for doc in nlp.pipe(textos, disable=["parser", "ner","lemmatizer"]):
        for token in doc:
            total_tokens+=1

            for token.text in ['¡','!']:
                total_exclamaciones +=1
            if not token.is_punct and not token.is_space:
                total_palabras+=1

                if token.is_upper and len(token.text)>1:
                    total_mayusculas+=1        

    pct_mayusculas=(total_mayusculas/total_palabras)*100 if total_palabras >0 else 0
    pct_exclamaciones=(total_exclamaciones/total_tokens)*100 if total_tokens >0 else 0
   

    return {
        "% Palabras Mayúsculas": pct_mayusculas,
        "% Exclamaciones": pct_exclamaciones
            }
                    
```

```python

def clasificar_odio_dataframe(df):
   
    # lista de lemas tóxicos 
    lemas_toxicos = [
        'puto', 'puta', 'gentuza', 'mierda', 'basura', 'idiota', 'tonto','miserable',
        'imbécil', 'cabrón', 'gilipollas', 'facha', 'ladrón', 'asqueroso','hdlgp','maldito' ]
    
    # Columna a analizar es de tipo texto
    textos = df['CONTENIDO A ANALIZAR'].astype(str).tolist()
    scores = []
    etiquetas = []
    
    # Procesamiento NLP optimizado
    for doc in nlp.pipe(textos, disable=["ner", "parser"]):
        #Creo un contador de score, insultos, palabras mayusculas, signos de exclamaciones y un booleano para determinar si hay o no un pronombre en el documento
        score = 0
        insulto=0
        mayuscula=0
        exclamaciones=0
        hay_pronombre = False

        #Si el documento está vacío, entonces no tiene score y lo etiqueto como No ODio
        if len(doc) == 0:
            scores.append(0)
            etiquetas.append("NO ODIO")
            continue
        #Si el lema está en la lista de lemas tóxicos predefinido, entonces incremento el contador de insultos    
        for token in doc:
            # Léxico Tóxico Lematizado (Suma 30 pts)
            if token.lemma_.lower() in lemas_toxicos:
                insulto+=1
                
            # Detección de ataques directos mediante pronombres (Suma 10 pts)
            if token.pos_ == "PRON":
                hay_pronombre = True
                
                
            # SI la palabra está en maypuscula ,incremento el contador de mayúsculas
            if token.is_upper and len(token.text) > 1:
                mayuscula+=1
                
            # Igual lógica para los signos de exclamación con el contador exclamaciones
            if token.text in ['¡', '!']:
                exclamaciones +=1
                
        # Definición de pesos 
        score +=(insulto * 40)
        if mayuscula >1:
            score+=15
        if exclamaciones >0:
            score +=10
        if insulto>0 and hay_pronombre:
            score+=20
            
        score_final = min(score, 100)
        scores.append(score_final)
        
        # Clasificación final
        if score_final > 50:
            etiquetas.append("ODIO")
        else:
            etiquetas.append("NO ODIO")
            
    # Añadimos los resultados al DataFrame original
    df_resultado = df.copy()
    df_resultado['Hate_Score'] = scores
    df_resultado['Clasificacion_Final'] = etiquetas
    
    return df_resultado


```
